```python
raw = [
    {"well_id": " w-101 ", "date": "2026-11-01", "oil_bbl": "120.5"},
    {"well_id": "W-102",   "date": "2026-11-01", "oil_bbl": "98"},
    {"well_id": "w-101",   "date": "2026-11-02", "oil_bbl": "115.0"},
    {"well_id": "W-103",   "date": "2026-11-01", "oil_bbl": ""},
    {"well_id": "W-102",   "date": "2026-11-02", "oil_bbl": "-5"},
    {"well_id": "W-103",   "date": "2026-11-02", "oil_bbl": "80.25"},
    {"well_id": "W-102",   "date": "2026-11-03", "oil_bbl": "101.5"},
    {"well_id": "W-104",   "date": "2026-11-03", "oil_bbl": "n/a"},
]
```

```python
def to_number(text: str) -> float | None:
    """This function converts a valid text/string number to a float. If not valid, then None"""
    text = text.strip()
    if text.replace(".", "", 1).isdigit():
        return float(text)
    elif text.startswith("-") and text.replace("-", "", 1).replace(".", "",1).isdigit():
        return float(text)
    else:
        return None
    
def clean_well(well_name: str) -> str:
    """This function trims well_name and uppercase it afterward"""
    return well_name.strip().upper()

def clean_record(rec: list):
    """Return a cleaned copy raw production records"""
    first_clean = []

    for data in rec:
        clean_dict = {}
        for key, value in data.items():
            if key == "oil_bbl":
                clean_value = to_number(value)
                clean_dict[key] = clean_value
            elif key == "well_id":
                clean_name = clean_well(value)
                clean_dict[key] = clean_name
            else:
                clean_dict[key] = value
        first_clean.append(clean_dict)
    
    return first_clean
```

```python
clean_data = clean_record(raw)
clean_data
```

```python
valid = []
rejects = []

for data in clean_data:
    if data["oil_bbl"] == None:
        reason = "missing/invalid oil"
        data["reasons"] = reason
        rejects.append(data)
    elif data['oil_bbl'] < 0:
        reason = "negative oil"
        data["reasons"] = reason
        rejects.append(data)
    else:
        valid.append(data)
```
