### Tasks

1. Write `to_number(text)`. It returns a `float`, or `None` if the text is empty or not a number, and it must handle a leading minus sign. Don't use try/except yet; that comes on Day 30.
1. Write `clean_record(rec)`. It strips and uppercases well_id and converts oil_bbl with to_number.
1. Loop through the records and split them into valid and rejects. Give each reject a reason: "missing/invalid oil" or "negative oil".
1. Build a dictionary of total oil per well from the valid records.
1. Write `summarize(...)`. It returns a dict with records read, valid, rejected, total oil and top well.
1. Print a tidy report with f-strings.
1. Bonus: run `display(spark.createDataFrame(valid))`. That's your first Spark DataFrame.

```python
raw = [
    {"well_id": " w-101 ", "date": "2026-11-01", "oil_bbl": "120.5"},
    {"well_id": "W-102",   "date": "2026-11-01", "oil_bbl": "98"},
    {"well_id": "w-101",   "date": "2026-11-02", "oil_bbl": "115.0"},
    {"well_id": "W-103",   "date": "2026-11-01", "oil_bbl": ""},
    {"well_id": "W-102",   "date": "2026-11-02", "oil_bbl": "-5"},
    {"well_id": "W-103",   "date": "2026-11-02", "oil_bbl": "80.25"},
    {"well_id": "W-102",   "date": "2026-11-03", "oil_bbl": "101.5"},
    {"well_id": "W-104",   "date": "2026-11-03", "oil_bbl": "n/a"},
]
```

```python
def to_number(text: str) -> float | None:
    """Converts a valid text/string number to a float. If not valid, then None"""
    text = text.strip()
    if text.replace(".", "", 1).isdigit():
        return float(text)
    elif text.startswith("-") and text.replace("-", "", 1).replace(".", "",1).isdigit():
        return float(text)
    else:
        return None
    
def clean_well(well_name: str) -> str:
    """Trims well_name and uppercase it afterward"""
    return well_name.strip().upper()

def clean_record(rec: dict) -> dict:
    """Return a cleaned copy of one raw production record"""    
    return {
        "well_id": clean_well(rec["well_id"]),
        "date": rec["date"],
        "oil_bbl": to_number(rec["oil_bbl"])
    }
```

```python
clean_data = []
for rec in raw:
    clean_data.append(clean_record(rec))

clean_data
```

```python
valid = []
rejects = []

for data in clean_data:
    if data["oil_bbl"] is None:
        reject = dict(data)
        reject["reason"] = "missing/invalid oil"
        rejects.append(reject)
    elif data["oil_bbl"] < 0:
        reject = dict(data)
        reject["reason"] = "negative oil"
        rejects.append(reject)
    else:
        valid.append(data)
```

```python
def total_oil_by_well(records: list[dict]) -> dict[str, float]:
    """Return total oil per well"""
    oil_by_well = {}
    for rec in records:
        well = rec["well_id"]
        if well not in oil_by_well:
            oil_by_well[well] = rec["oil_bbl"]
        else:
            oil_by_well[well] += rec["oil_bbl"]

    return oil_by_well

oil_per_well = total_oil_by_well(valid)
oil_per_well
```

```python
def summarize(
    raw_data: list[dict],
    valid_data: list[dict],
    rejected_data: list[dict],
    oil_by_well: dict[str, float]
) -> dict:
    """Gives a dictionary summary of raw well data and processed data details"""
    top_well = None
    top_oil = 0
    for key, value in oil_by_well.items():
        if value > top_oil:
            top_oil = value
            top_well = key
    return {
        "records_read": len(raw_data),
        "valid": len(valid_data),
        "rejected": len(rejected_data),
        "total_oil_bbl": round(sum(oil_by_well.values()), 2),
        "top_well": top_well,
    }

summarize_report = summarize(raw_data=raw, valid_data=valid, rejected_data=rejects, oil_by_well=oil_per_well)

summarize_report
```

```python
def print_report(summary, oil_by_well, rejected_data)-> str:
    """Outputs a detailed summary of the well data"""
    print("=== Well Production Report ===")
    print(f"{'Records read:':<16}{summary["records_read"]}")
    print(f"{'Valid:':<16}{summary["valid"]}")
    print(f"{'Rejected:':<16}{summary["rejected"]}\n")
    print("Oil by well (bbl):")
    for key, value in oil_by_well.items():
        print(f"  {key:<8} {value:10.2f}")
    print(" ","-" * 20)
    print(f"  {'Total':<8} {summary["total_oil_bbl"]:10.2f}\n")
    print(f"Top well: {summary["top_well"]}\n")
    print("Rejects:")
    for rec in rejected_data:
        print(f"  {rec["well_id"]} {rec["date"]} {rec["reason"]}")

print_report(summary=summarize_report, oil_by_well=oil_per_well, rejected_data=rejects)
```
